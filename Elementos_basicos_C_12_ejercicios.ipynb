## 12 — Reto final: encuentra, corrige y modifica

Depura el programa de compra. Después agrega un impuesto del 16% y muestra subtotal, impuesto y total. Finalmente responde: ¿para qué sirve &, por qué total es double, qué hace %29s y qué ocurre si falta el ;?


```python
%%script bash
cat << 'EOF' > ejercicio12.c

#include <stdio.h>

int main(void)
{
    char producto[30];
    int cantidad;
    double precio, total;

    printf("Producto: ");
    scanf("%29s", producto);

    printf("Precio: ");
    scanf("%lf", &precio);

    printf("Cantidad: ");
    scanf("%d", cantidad);

    total = precio * cantidad

    printf("\nProducto: %s\n", producto);
    printf("Total: %.2f\n", total);

    return 0;
}
EOF
gcc ejercicio12.c -o ejercicio12
```
